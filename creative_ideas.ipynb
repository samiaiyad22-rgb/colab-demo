# Creative Ideas with Everyday Objects

This notebook asks for one everyday object and uses the OpenAI Responses API to suggest **one unusual but practical use**.

Run each cell from top to bottom. You will enter your API key at a hidden prompt and then type an object, such as a paper clip. API use may incur charges, and your account needs access to `gpt-6-luna`.

## 1. Check for the OpenAI Python SDK

This cell installs the SDK in the Colab runtime only if it is not already available.

```python
# Check whether the OpenAI package is already installed.
import importlib.util
import subprocess
import sys

if importlib.util.find_spec("openai") is None:
    # Install the package quietly into this Colab runtime.
    subprocess.check_call(
        [sys.executable, "-m", "pip", "install", "--quiet", "openai"]
    )

print("The OpenAI SDK is ready.")
```

## 2. Enter your API key

`getpass` hides the key while you type. The key is passed directly to the OpenAI client; it is not printed or written into this notebook. Do not put it in a code cell.

```python
from getpass import getpass
from openai import OpenAI

# The input stays hidden and is not displayed in the cell output.
client = OpenAI(api_key=getpass("Enter your OpenAI API key: "))
print("The API client is ready.")
```

## 3. Enter an everyday object

Choose one familiar object. The notebook will make one suggestion for it.

```python
# Ask for one object and remove extra spaces around the answer.
everyday_object = input("What everyday object should I use? ").strip()

if not everyday_object:
    raise ValueError("Please enter an object before continuing.")

print(f"Object: {everyday_object}")
```

## 4. Get one creative idea

The request asks for one safe, realistic use and tells the model not to list alternatives.

```python
# Ask for exactly one unusual but practical use.
prompt = (
    f"Everyday object: {everyday_object}\n\n"
    "Suggest exactly ONE unusual but practical use for this object. "
    "Give it a short title and explain how to use it in 2-3 sentences. "
    "Keep the idea safe and realistic. Do not list alternatives."
)

response = client.responses.create(
    model="gpt-6-luna",
    input=prompt,
)

# Display only the model's text response.
print(response.output_text)
```
